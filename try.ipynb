```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import os

# 1. Définissez l'identifiant du modèle et le dossier de destination local
model_id = "google/medgemma-1.5-4b-it"
local_dir = "./medgemma" # Le dossier où les fichiers seront sauvegardés

print("Téléchargement du tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.save_pretrained(local_dir)

print("Téléchargement du modèle (cela peut prendre quelques minutes selon votre connexion)...")
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto", # Utilise le GPU si disponible, sinon le CPU
    torch_dtype="auto" # Optimise la mémoire (float16 ou bfloat16 si supporté)
)
model.save_pretrained(local_dir)

print(f"Modèle téléchargé et sauvegardé avec succès dans le dossier : {os.path.abspath(local_dir)}")
```
